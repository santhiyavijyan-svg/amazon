```python
!pip -q install tensorflow scikit-learn

import re
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from google.colab import files
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error, mean_squared_error, accuracy_score, classification_report
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras import layers, Model

np.random.seed(42)
tf.random.set_seed(42)
print("TensorFlow:", tf.__version__)
```

```python
RATING_COL = None  # Example: "rating" or "reviews.rating"
REVIEW_COL = None  # Example: "review" or "review_content"
PRODUCT_COL = None # Optional, e.g. "product_name"
CATEGORY_COL = None # Optional, e.g. "category"
PRICE_COL = None   # Optional, e.g. "price"

def norm_col(c):
    return re.sub(r"[^a-z0-9]+", "_", str(c).strip().lower()).strip("_")

normalized = {c: norm_col(c) for c in df.columns}

def find_col(explicit, candidates, contains_any=()):
    if explicit is not None:
        if explicit not in df.columns:
            raise ValueError(f"Configured column '{explicit}' not found. Columns: {df.columns.tolist()}")
        return explicit
    for c, n in normalized.items():
        if n in candidates:
            return c
    for c, n in normalized.items():
        if any(token in n for token in contains_any):
            return c
    return None

RATING_COL = find_col(RATING_COL,
    {"rating","ratings","star_rating","stars","overall","review_rating","reviews_rating","rating_score"},
    ("rating","stars","overall"))
REVIEW_COL = find_col(REVIEW_COL,
    {"review","reviews","review_text","review_content","text","comment","comments","customer_review","summary"},
    ("review","comment","review_text"))
PRODUCT_COL = find_col(PRODUCT_COL,
    {"product_name","product_title","title","name","product"},
    ("product_name","product_title"))
CATEGORY_COL = find_col(CATEGORY_COL,
    {"category","product_category","main_category"},
    ("category",))
PRICE_COL = find_col(PRICE_COL,
    {"price","product_price","discounted_price","actual_price"},
    ("price",))

print("Detected rating column:", RATING_COL)
print("Detected review column:", REVIEW_COL)
print("Detected product column:", PRODUCT_COL)
print("Detected category column:", CATEGORY_COL)
print("Detected price column:", PRICE_COL)

if RATING_COL is None or REVIEW_COL is None:
    raise ValueError(
        "Could not detect rating/review columns automatically. "
        "Set RATING_COL and REVIEW_COL manually in this cell, using names printed above."
    )
```

```python
df = df.drop_duplicates().copy()
df[RATING_COL] = pd.to_numeric(df[RATING_COL], errors="coerce")
df = df.dropna(subset=[RATING_COL, REVIEW_COL]).copy()

# Standardize review text and rating values
df[REVIEW_COL] = df[REVIEW_COL].fillna("").astype(str)
df = df[df[RATING_COL].between(1, 5)].copy()

# Round ratings to the nearest star if source contains decimal ratings
df["target_rating"] = df[RATING_COL].round().astype(int)
df["clean_review"] = (
    df[REVIEW_COL].astype(str).str.lower()
    .str.replace(r"https?://\\S+|www\\.\\S+", " ", regex=True)
    .str.replace(r"[^a-z0-9\\s]", " ", regex=True)
    .str.replace(r"\\s+", " ", regex=True).str.strip()
)

# Product text fields: fill missing values and normalize
text_product_cols = [c for c in [PRODUCT_COL, CATEGORY_COL] if c is not None]
for c in text_product_cols:
    df[c] = df[c].fillna("Unknown").astype(str).str.lower().str.strip()

if PRICE_COL is not None:
    df[PRICE_COL] = pd.to_numeric(
        df[PRICE_COL].astype(str).str.replace(r"[^0-9.]", "", regex=True),
        errors="coerce"
    )
    df[PRICE_COL] = df[PRICE_COL].fillna(df[PRICE_COL].median())

# If the review became empty after cleaning, keep it as an empty string.
df["clean_review"] = df["clean_review"].fillna("")
df = df.reset_index(drop=True)

print("Rows after cleaning:", len(df))
print("Missing values in selected fields:")
display(df[[RATING_COL, REVIEW_COL, "clean_review", "target_rating"]].isnull().sum().to_frame("missing"))
display(df[[REVIEW_COL, "clean_review", "target_rating"]].head())
print("Rating distribution:")
display(df["target_rating"].value_counts().sort_index().to_frame("count"))
```

```python
MAX_WORDS = 10000
MAX_LEN = 120

tokenizer = Tokenizer(num_words=MAX_WORDS, oov_token="<OOV>")
tokenizer.fit_on_texts(df["clean_review"].tolist())

sequences = tokenizer.texts_to_sequences(df["clean_review"].tolist())
X_text = pad_sequences(sequences, maxlen=MAX_LEN, padding="post", truncating="post")
y = df["target_rating"].to_numpy(dtype=np.float32)

print("Vocabulary size:", min(MAX_WORDS, len(tokenizer.word_index) + 1))
print("Encoded review matrix shape:", X_text.shape)
print("Example cleaned review:", df["clean_review"].iloc[0])
print("Example numerical sequence:", X_text[0][:30])
```

```python
numeric_cols = [c for c in [PRICE_COL] if c is not None]
categorical_cols = [c for c in [PRODUCT_COL, CATEGORY_COL] if c is not None]

# Avoid target leakage if selected columns overlap with rating/review columns.
numeric_cols = [c for c in numeric_cols if c not in {RATING_COL, REVIEW_COL}]
categorical_cols = [c for c in categorical_cols if c not in {RATING_COL, REVIEW_COL}]

# Split indices first so product preprocessing is fitted only on training data.
indices = np.arange(len(df))
train_idx, test_idx = train_test_split(
    indices, test_size=0.2, random_state=42, stratify=df["target_rating"]
)

# Use a separate validation subset from training data.
train_idx, val_idx = train_test_split(
    train_idx, test_size=0.2, random_state=42, stratify=df.loc[train_idx, "target_rating"]
)

# Preprocess structured product information
preprocessor = None
if numeric_cols or categorical_cols:
    transformers = []
    if numeric_cols:
        transformers.append(("num", Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]), numeric_cols))
    if categorical_cols:
        transformers.append(("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical_cols))
    preprocessor = ColumnTransformer(transformers=transformers, remainder="drop")
    X_product_train = preprocessor.fit_transform(df.iloc[train_idx])
    X_product_val = preprocessor.transform(df.iloc[val_idx])
    X_product_test = preprocessor.transform(df.iloc[test_idx])
    # Convert sparse matrices to dense float32 for Keras; suitable for small datasets.
    if hasattr(X_product_train, "toarray"):
        X_product_train = X_product_train.toarray()
        X_product_val = X_product_val.toarray()
        X_product_test = X_product_test.toarray()
    X_product_train = np.asarray(X_product_train, dtype=np.float32)
    X_product_val = np.asarray(X_product_val, dtype=np.float32)
    X_product_test = np.asarray(X_product_test, dtype=np.float32)
else:
    X_product_train = np.zeros((len(train_idx), 1), dtype=np.float32)
    X_product_val = np.zeros((len(val_idx), 1), dtype=np.float32)
    X_product_test = np.zeros((len(test_idx), 1), dtype=np.float32)

X_text_train, X_text_val, X_text_test = X_text[train_idx], X_text[val_idx], X_text[test_idx]
y_train, y_val, y_test = y[train_idx], y[val_idx], y[test_idx]

print("Training rows:", len(train_idx), "Validation rows:", len(val_idx), "Test rows:", len(test_idx))
print("Product feature count:", X_product_train.shape[1])
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

# 1. Text Input Pathway
text_input = layers.Input(shape=(MAX_LEN,), name="text_input")
embedding = layers.Embedding(input_dim=MAX_WORDS, output_dim=64, name="text_embedding")(text_input)
global_pooling = layers.GlobalAveragePooling1D(name="text_pooling")(embedding)
text_dense = layers.Dense(32, activation="relu")(global_pooling)

# 2. Product Structured Features Pathway
product_feature_dim = X_product_train.shape[1]
product_input = layers.Input(shape=(product_feature_dim,), name="product_input")
product_dense = layers.Dense(16, activation="relu")(product_input)

# 3. Concatenate Pathways & Output
concatenated = layers.concatenate([text_dense, product_dense])
dense_combined = layers.Dense(16, activation="relu")(concatenated)
dropout = layers.Dropout(0.2)(dense_combined)
output = layers.Dense(1, name="rating_output")(dropout)

# Build Model
model = Model(inputs=[text_input, product_input], outputs=output)

# Compile Model using Mean Squared Error as we are predicting numerical ratings
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

model.summary()
```

```python
early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss", patience=5, restore_best_weights=True
)

history = model.fit(
    [X_text_train, X_product_train], y_train,
    validation_data=([X_text_val, X_product_val], y_val),
    epochs=30,
    batch_size=16,
    callbacks=[early_stop],
    verbose=1
)

# Plot training and validation loss
plt.figure(figsize=(7, 4))
plt.plot(history.history["loss"], label="Training loss")
plt.plot(history.history["val_loss"], label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("MSE loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
pred_raw = model.predict([X_text_test, X_product_test], verbose=0).reshape(-1)
pred_rating = np.clip(pred_raw, 1, 5)
pred_rounded = np.clip(np.rint(pred_rating), 1, 5).astype(int)
actual_rating = y_test.astype(int)

mae = mean_absolute_error(actual_rating, pred_rating)
rmse = np.sqrt(mean_squared_error(actual_rating, pred_rating))
rounded_accuracy = accuracy_score(actual_rating, pred_rounded)

print(f"Test MAE: {mae:.3f} stars")
print(f"Test RMSE: {rmse:.3f} stars")
print(f"Rounded exact-match accuracy: {rounded_accuracy:.3f} ({rounded_accuracy*100:.1f}%)")
print("\\nClassification report (rounded star ratings):")
print(classification_report(actual_rating, pred_rounded, labels=[1,2,3,4,5], zero_division=0))
```

```python
unseen_reviews = [
    "Excellent product, works perfectly and the quality is amazing.",
    "Very poor quality. It stopped working after two days.",
    "The product is okay, not great but not bad either."
]

def make_product_row(product_name=None, category=None, price=None):
    # Use representative defaults for product features when no details are provided.
    row = {}
    for c in df.columns:
        if c == PRODUCT_COL:
            row[c] = product_name if product_name is not None else "unknown"
        elif c == CATEGORY_COL:
            row[c] = category if category is not None else "unknown"
        elif c == PRICE_COL:
            row[c] = price if price is not None else df[c].median()
        elif c == RATING_COL:
            row[c] = df[c].median()
        elif c == REVIEW_COL:
            row[c] = ""
        else:
            row[c] = df[c].iloc[0]
    return pd.DataFrame([row])

def predict_review(review, product_name=None, category=None, price=None):
    cleaned = re.sub(r"[^a-z0-9\\s]", " ", str(review).lower())
    cleaned = re.sub(r"\\s+", " ", cleaned).strip()
    seq = tokenizer.texts_to_sequences([cleaned])
    padded = pad_sequences(seq, maxlen=MAX_LEN, padding="post", truncating="post")

    if preprocessor is not None:
        row = make_product_row(product_name, category, price)
        product_features = preprocessor.transform(row)
        if hasattr(product_features, "toarray"):
            product_features = product_features.toarray()
        product_features = np.asarray(product_features, dtype=np.float32)
    else:
        product_features = np.zeros((1, 1), dtype=np.float32)

    raw = float(model.predict([padded, product_features], verbose=0)[0][0])
    clipped = float(np.clip(raw, 1, 5))
    star = int(np.clip(np.rint(clipped), 1, 5))
    return clipped, star

results = []
for review in unseen_reviews:
    score, stars = predict_review(review)
    results.append({
        "unseen_review": review,
        "predicted_rating": round(score, 2),
        "predicted_star_rating": stars
    })

display(pd.DataFrame(results))
```
